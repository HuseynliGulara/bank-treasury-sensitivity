```python
import pandas_datareader.data as web
import datetime

START = "2015-01-01"
END = "2026-09-04"

tickers = ["JPM", "BAC", "GS", "^GSPC"]
stock_prices = yf.download(tickers, start=START, end=END)["Close"]
stock_prices.to_csv("../data/raw/stock_prices.csv")
stock_prices.head()
```

```python
start = datetime.datetime.strptime(START, "%Y-%m-%d")
end = datetime.datetime.strptime(END, "%Y-%m-%d")
treasury = web.DataReader(["DGS10", "DGS2"], "fred", start, end)
treasury.to_csv("../data/raw/treasury_yields.csv")
treasury.head()
```
